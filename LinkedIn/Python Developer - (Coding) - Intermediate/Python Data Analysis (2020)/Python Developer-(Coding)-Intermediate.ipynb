**2. Data Structures in Pure Python**
===

Warmup with Python loops
---
- 📦 **Standard imports**: Common modules include NumPy (`np`), Pandas (`pd`), and Matplotlib (`pyplot`). Aliases make code shorter and clearer.  
- 📊 **Inline plotting**: Jupyter can display plots directly in the notebook with `%matplotlib inline`.  
- 🔁 **Loops in Python**:  
  - `for variable in iterable:` is the basic structure.  
  - Iterables include lists, dicts, and especially `range()`.  
  - `range(start, stop, step)` generates sequences; stop is exclusive.  
- 💡 **Nested loops example**: Breaking $1 into all coin combinations requires multiple nested loops. Python handles the nesting cleanly with indentation.  
- 🧮 **Refactoring strategy**:  
  - Loop by **value** instead of count (using step in `range`).  
  - Remove unnecessary loops (like pennies) by computing the remainder directly.  
- 🛠️ **Generalization**: Turn the code into a function (`find_combinations(total)`) to handle any amount, not just $1.  
- 📈 **Growth insight**: Plotting results shows the number of combinations grows roughly with the fifth power of the total.  
- 🎯 **Core lesson**: Loops are everywhere in data analysis. Mastering `for` and `range` is essential for building flexible, reusable solutions.

Sequences: Lists, tuples, and the slicing syntax
---
- 📋 **Lists basics**:  
  - Defined with square brackets `[]`, elements separated by commas.  
  - Can store mixed types (strings, numbers, other lists, etc.).  
  - Length obtained with `len()`.  
  - Indexed starting at **0**; negative indices count from the end.  

- ✏️ **List operations**:  
  - Reassign elements with bracket indexing.  
  - Add items: `append()` (single), `extend()` (multiple), `+` (concatenate).  
  - Insert at position: `insert()`.  
  - Remove items: `del` (by index), `remove()` (by value).  
  - Sort in place with `sort()`, or create a new sorted list with `sorted()`.  

- 🔄 **Slicing**:  
  - Syntax: `[start:end:step]`, with end exclusive.  
  - Can omit start/end for defaults, use negative indices, or steps for skipping.  
  - Useful for copying, reassigning, or deleting subsets.  
  - Same slicing rules apply later to NumPy arrays.  

- 🟦 **Empty list**: Written as `[]`, length zero.  

- 🟠 **Tuples**:  
  - Defined with parentheses `()`.  
  - Immutable (cannot modify or add elements).  
  - Support indexing and slicing like lists.  
  - Commonly used in **tuple unpacking** (assign multiple variables at once).  
  - Appear in iterators like `enumerate()` (looping over index + element).  
  - Can be unpacked into functions with `*tuple`.  

👉 In short: **Lists are flexible, mutable containers for data; tuples are immutable versions often used for unpacking and iteration.** 

Dictionaries and sets
---
- 📖 **Dictionaries (dicts)**:  
  - Defined with curly braces `{}` and key–value pairs (`"Italy": "Rome"`).  
  - Access values with bracket notation using keys, not indices.  
  - Add items by assigning to a new key; nonexistent keys raise `KeyError`.  
  - Check existence with `in`.  
  - Combine dicts using `update()` (handles duplicate keys by replacing values).  
  - Delete items with `del`.  
  - Keys must be **hashable objects** (e.g., strings, numbers, tuples).  

- 🔄 **Looping over dicts**:  
  - `for key in dict` → iterate over keys.  
  - `dict.keys()` → explicit keys iterator.  
  - `dict.values()` → iterate over values.  
  - `dict.items()` → iterate over key–value pairs (often with tuple unpacking).  
  - Since Python 3.7, dicts preserve **insertion order**.  

- 🟦 **Empty dict**: Written as `{}`, length zero.  

- 🟢 **Sets**:  
  - Defined with braces `{}` but without colons.  
  - Store unique items (no duplicates).  
  - Support membership checks (`in`), adding/removing items, and looping.  
  - Do not support indexing.  

- ⚡ **Importance**: Dicts and sets are fundamental in Python—dicts underpin class attributes, methods, and many third-party packages (like Pandas).  

👉 In short: **Dicts map keys to values with flexible looping and ordering; sets store unique items without duplicates. Both are core building blocks in Python.**  

Comprehensions
---
- 📝 **Comprehensions = Pythonic loops**: They provide a concise, readable way to build lists, dicts, or sets compared to traditional loops.  
- 📋 **List comprehensions**:  
  - Syntax: `[expression for item in iterable]`.  
  - Can include an `if` clause to filter results.  
  - Example: `[x**2 for x in range(10) if x % 4 == 0]`.  
- 📚 **Dict comprehensions**:  
  - Syntax: `{key: value for item in iterable}`.  
  - Useful for building mappings or inverting existing dicts.  
- ⚡ **Generator expressions**:  
  - Similar syntax but without brackets.  
  - Produce values one at a time, saving memory (e.g., `sum(x**2 for x in range(10))`).  
- 🔄 **Nested comprehensions**: Multiple loops can be written in sequence inside a comprehension, enabling compact solutions for complex iterations.  
- 🎯 **Practical benefit**: Comprehensions replace older constructs like `map()` and `filter()` in Python, making code more expressive and efficient.  
- 🚀 **Core lesson**: Comprehensions are a powerful tool for manipulating lists, dicts, and data—essential for clean, efficient Python data analysis.

Advanced Python containers
---
- 🟠 **Tuples**:  
  - Immutable sequences (cannot be changed after creation).  
  - Useful for storing records like `(first_name, last_name, birthday)`.  
  - Access fields by index, but this can reduce readability and increase risk of bugs.  

- 🏷️ **Named Tuples (`collections.namedtuple`)**:  
  - Tuples with **field names** for clarity (`person.first_name` instead of `person[0]`).  
  - Can be created with positional or keyword arguments.  
  - Support both index-based and name-based access.  
  - Improve code readability and expressiveness.  

- 📦 **Data Classes (`dataclasses.dataclass`)**:  
  - Introduced in Python 3.7 (available as a package in 3.6).  
  - Full Python classes with fields, default values, and methods.  
  - Fields accessed by name only.  
  - Support advanced features: freezing (immutability), ordering, inheritance, and custom methods.  
  - More powerful than named tuples, especially for object-oriented programming.  

- 🗂️ **Default Dictionaries (`collections.defaultdict`)**:  
  - Provide default values for missing keys.  
  - Avoid repetitive code when building dicts of lists or other containers.  
  - Example: `defaultdict(list)` automatically creates an empty list for new keys.  
  - Useful for grouping data (e.g., people by birthday).  

👉 In short:  
- **Tuples** = simple, immutable records.  
- **Named Tuples** = tuples with field names for clarity.  
- **Data Classes** = full-featured, flexible records with methods.  
- **Defaultdict** = smarter dicts that handle missing keys gracefully.

**3. Wordplays: Anagrams and Palindromes**
===

Anagrams overview
---
- 🔤 **Definition of anagram**: Two words are anagrams if their letters can be rearranged to form each other (e.g., *elvis* ↔ *lives*).  
- 📝 **Signature strategy**:  
  - Define a word’s signature as its letters sorted alphabetically (e.g., *post* → `opst`).  
  - Words with the same signature are anagrams of each other (*spot, stop, tops, pots, opts*).  
- 📂 **Data structure**:  
  - Use a **Python dictionary** where keys = signatures, values = lists of words.  
  - This makes lookup efficient: compute a word’s signature, check if it exists in the dict.  
- ⚙️ **Implementation plan**:  
  1. Read all words from the dictionary.  
  2. Compute each word’s signature.  
  3. Store the word under its signature in the dict.  
  4. To find anagrams, look up the signature.  
- 🎯 **Outcome**: A simple, practical project that applies loops, dicts, and comprehensions to solve a real-world problem.  

👉 In short: By indexing words by their **sorted-letter signature**, Python can quickly group and retrieve anagrams from a dictionary.

Loading a dictionary
---
- 📂 **Source file**: The exercise files include `words.txt`, a 1934 UNIX dictionary with ~200,000 words. You can also substitute another dictionary file if desired.  
- 📖 **Python idiom**: The preferred way to read a text file is to open it in read mode (`"r"`) and loop over its lines directly.  
- 📝 **Initial cleanup**:  
  - Words contain trailing newline characters (`\n`) → remove with `.strip()`.  
  - Words may be capitalized → normalize with `.lower()`.  
- 🔄 **Duplicates**: Using a **list** will keep duplicates (e.g., “A” vs “a”). Switching to a **set** automatically removes duplicates.  
- ⚡ **Comprehension style**: Instead of a loop, use a set comprehension for a more Pythonic one-liner:  
  ```python
  words = {line.strip().lower() for line in open("words.txt", "r")}
  ```  
- 🔠 **Sorting**: Wrap the set in `sorted()` to restore alphabetical order while keeping uniqueness.  
- 🌍 **Unicode support**: Python 3 strings are natively Unicode, so they handle international dictionaries (e.g., French).  
- 🔑 **Encoding awareness**: When reading non-English files, specify the correct encoding (e.g., ISO-8859-1 / Latin-1 for French).  

👉 In short: **Load words → clean with `strip()` + `lower()` → store in a set for uniqueness → optionally sort → ready for anagram analysis.** 

Finding anagrams
---
- 📝 **Signature function**:  
  - A word’s signature is its letters sorted alphabetically, joined back into a string.  
  - Example: `"post"` → `"opst"`.  
  - This allows quick comparison of anagram candidates.  

- ⚡ **Brute-force vs optimized**:  
  - Brute-force: compute signature for every word and compare → works but slow for large lists.  
  - Optimized: build a dictionary (`words_by_sig`) where keys = signatures, values = sets of words.  

- 🗂️ **Using `defaultdict`**:  
  - Perfect for grouping words by signature.  
  - Each new signature automatically starts with an empty set.  
  - Example:  
    ```python
    from collections import defaultdict
    words_by_sig = defaultdict(set)
    for word in words:
        sig = ''.join(sorted(word))
        words_by_sig[sig].add(word)
    ```

- 🔍 **Filtering**:  
  - Remove trivial cases (signatures with only one word).  
  - Achieved with a dictionary comprehension:  
    ```python
    anagrams_by_sig = {sig: ws for sig, ws in words_by_sig.items() if len(ws) > 1}
    ```

- 🛡️ **Safe lookup**:  
  - Function `find_anagram(word)` computes signature and looks up in dict.  
  - Use `try/except KeyError` to return an empty set if no anagram exists.  

- 📊 **Explorations**:  
  - Find longest words with anagrams → sort by signature length.  
  - Find largest anagram groups → sort by group size.  
  - Example: longest anagrams had **22 letters**, largest groups had **10 words**.  

👉 In short: By indexing words by their **sorted-letter signature** in a `defaultdict(set)`, Python can efficiently group and retrieve anagrams, enabling further analysis like longest or largest anagram sets.

Challenge: Palindromes
---
### 🔑 Key idea
- A **palindromic pair** occurs when one word equals the reverse of another (e.g., *reward ↔ drawer*).  
- A **true palindrome** is when the word equals its own reverse (e.g., *radar*).  
- Python slicing makes reversal easy: `word[::-1]`.

### 🛠️ Implementation steps
1. **Load words** (already cleaned and deduplicated from `words.txt`).  
2. **Build a set** of words for fast membership checks.  
3. **Loop through each word**:  
   - Compute its reverse with `[::-1]`.  
   - If the reverse exists in the set, record the pair.  
4. **Avoid duplicates**: Only store pairs once (e.g., `("reward", "drawer")` but not both orders).  
5. **Include palindromes**: Words equal to their reverse can be collected separately.

### 🚀 What you’ll get
- **Pairs** like `("reward", "drawer")`, `("stressed", "desserts")`.  
- **True palindromes** like `radar`, `level`, `madam`.  

👉 This challenge nicely shows how **sets + slicing** make wordplay problems efficient in Python.

````python
# ✅ Example code

# Load words (already cleaned and deduplicated)
with open("words.txt", "r", encoding="utf-8") as f:
    words = {line.strip().lower() for line in f}

palindromic_pairs = set()
palindromes = set()

for word in words:
    rev = word[::-1]
    if word == rev:
        palindromes.add(word)  # true palindrome
    elif rev in words and word < rev:  # avoid duplicates
        palindromic_pairs.add((word, rev))

print("Some palindromic pairs:", list(palindromic_pairs)[:10])
print("Some true palindromes:", list(palindromes)[:10])
````

Solution: Palindromes
---
- 🔄 **Reversal with slicing**: In Python, reversing a string is done with `word[::-1]`. This uses slicing with a negative step.  
- 🧩 **Connection to anagrams**: Palindromic pairs are also anagrams, since reversing preserves the set of letters.  
- 🗂️ **Strategy**:  
  - Loop over each set of words grouped by signature (from the anagram dictionary).  
  - For each pair of words in the set, check if one equals the reverse of the other.  
- ⚙️ **Avoiding duplicates**:  
  - Either manually ensure pairs are only checked once (e.g., by sorting them before comparison).  
  - Or use `itertools.combinations` to generate unique pairs cleanly.  
- ✅ **Result**:  
  - Captures both **palindromic pairs** (e.g., *reward ↔ drawer*, *stressed ↔ desserts*) and **true palindromes** (e.g., *radar*, *level*).  
  - The `itertools.combinations` approach makes the code shorter, more expressive, and avoids redundant checks.  

👉 In short: The challenge demonstrates how **string slicing + dict grouping + itertools** combine to elegantly find palindromic word pairs in a dictionary.

**4. Arrays with NumPy**
===

NumPy overview
---
- ⚡ **Why NumPy?**  
  - Provides **multi-dimensional arrays** that are fast and memory-efficient.  
  - Essential for handling **large datasets** and mathematical applications.  
  - A core package in the Python data analysis ecosystem.  

- 🧩 **Difference from Python containers**:  
  - Python variables are **labels** pointing to objects in memory.  
  - Lists and dicts can hold heterogeneous elements, but this flexibility is inefficient for large numeric data.  
  - NumPy arrays store values **side by side in contiguous memory**, all of the same type → faster and more efficient.  

- 🔢 **Array structure**:  
  - 1D arrays: indexed by a single integer.  
  - 2D arrays: indexed by two integers (rows, columns).  
  - Indices start at **0** and go up to **N-1**.  

- 🛠️ **Data types in NumPy**:  
  - More precise than Python’s built-in types.  
  - Integers: `int8`, `int16`, `int32`, `int64` (and unsigned versions).  
  - Floats: `float16`, `float32`, `float64`, `float128` (platform-dependent).  
  - Others: complex numbers, booleans, bytes, Unicode strings, void (record arrays), and object (pointer to Python objects).  
  - Most common types: `int64` and `float64`.  

👉 In short: NumPy arrays are **specialized, efficient containers** for homogeneous data, designed for speed, memory efficiency, and interoperability with lower-level languages like C and Fortran.

Creating NumPy arrays
---
- 📂 **Loading arrays from files**:  
  - `np.loadtxt("monalisa.txt")` → loads text data into a 2D array.  
  - `np.load("monalisa.npy")` → loads NumPy’s native binary format (cross-platform, faster).  
  - Arrays can be queried for **dimensions**, **shape**, **size**, and **dtype**.  

- 🎨 **Visualizing arrays**:  
  - Use `matplotlib.pyplot.imshow()` to display 2D arrays as images.  
  - Apply colormaps like `"gray"` for grayscale.  
  - 3D arrays can represent RGB images directly.  
  - Adjust figure size with `plt.figure(figsize=(w,h))`.  

- 🛠️ **Creating arrays manually**:  
  - From lists: `np.array([1,2,3])`.  
  - Empty arrays: `np.empty(shape, dtype)`.  
  - Zeros: `np.zeros(shape)` or `np.zeros_like(other_array)`.  
  - Regular sequences:  
    - `np.linspace(start, stop, num)` → evenly spaced values.  
    - `np.arange(start, stop, step)` → like Python’s `range()`.  
  - Random arrays:  
    - `np.random.rand(shape)` → uniform [0,1].  
    - `np.random.randint(low, high, size)` → random integers.  
    - `np.random.randn(size)` → normal distribution.  

- 💾 **Saving arrays**:  
  - `np.save("filename.npy", array)` → binary format.  
  - `np.savetxt("filename.txt", array)` → readable ASCII text.  
  - Reload with `np.load` or `np.loadtxt`.  

👉 In short: NumPy arrays can be **loaded, created, visualized, and saved** with simple functions, making them the backbone of numerical computing in Python.

Indexing NumPy arrays
---
- 🔢 **Indexing basics**:  
  - Use multiple indices inside brackets: `array[row, col, channel]`.  
  - Negative indices count back from the end (like Python lists).  
  - Indexing beyond boundaries raises an `IndexError`.  
  - You can assign values directly to indexed elements.  

- 📐 **Slicing arrays**:  
  - Works like Python lists but extended to multiple dimensions: `array[400:800, 200:600]`.  
  - Use `:` for full ranges, and `...` (ellipsis) for multiple full slices.  
  - Steps reduce resolution: `array[::20]` selects every 20th pixel.  
  - Backward slicing (`[::-1]`) reverses rows or columns.  
  - Mixing slicing and indexing reduces dimensionality (e.g., fixing one index yields a vector).  

- 🎨 **Bulk modifications**:  
  - Assigning to slices modifies large sections at once (e.g., filling a square with zeros).  
  - Matching slices on both sides allows replacement with another array (e.g., random pixels).  

- 🧩 **Fancy indexing**:  
  - Use arrays (often Boolean masks) to select subsets of elements.  
  - Example: thresholding an image by selecting pixels darker than a value.  

- ⚡ **Important difference from lists**:  
  - **Python lists**: slicing creates a **copy**.  
  - **NumPy arrays**: slicing creates a **view** pointing to the same memory.  
  - Modifying the slice also modifies the original array.  
  - To make an independent copy, use `.copy()`.  

👉 In short: NumPy indexing and slicing are powerful, allowing precise element access, efficient bulk operations, and advanced techniques like Boolean masking. Unlike lists, slices are **views**, not copies—so changes propagate unless you explicitly copy.  

Doing math with NumPy arrays
---
- 🧮 **Universal functions (ufuncs)**:  
  - NumPy provides element‑wise functions like `np.sin`, `np.cos`, `np.log`.  
  - They operate directly on arrays, returning arrays of the same shape.  
  - This makes it easy to plot mathematical functions with Matplotlib.  

- 📊 **Plotting multiple functions**:  
  - Use `plt.plot(x, y)` for each function.  
  - Matplotlib automatically cycles colors and can add legends with `plt.legend()`.  

- ➕ **Array operations**:  
  - Operations between arrays of the same shape work element‑wise.  
  - Example: `np.sin(x) + np.cos(x)`.  
  - Different shapes usually fail, except when **broadcasting** applies.  

- 📡 **Broadcasting**:  
  - NumPy can align arrays of different dimensions if compatible.  
  - Example: adding a scalar to an array (`array + 1.5`).  
  - Example: multiplying a 2D image by a 1D vector → horizontal gradient.  
  - To apply along rows, reshape with `np.newaxis` to add a dimension → vertical gradient.  

- 🎨 **Practical demo**:  
  - Broadcasting lets you apply gradients or transformations across images efficiently.  
  - Example: Mona Lisa image multiplied by row/column vectors to create shading effects.  

- ⚡ **Advanced operations**:  
  - NumPy supports fast Fourier transforms, random numbers, statistics, interpolation, and linear algebra.  
  - Since Python 3.5, the `@` operator is available for matrix multiplication.  
    - Example: `A @ B` is equivalent to `np.dot(A, B)`.  

👉 In short: NumPy arrays + ufuncs + broadcasting = **powerful, efficient numerical computing**. You can apply mathematical functions to entire datasets, combine arrays, and manipulate images or matrices with minimal code.

Special arrays: Records and dates
---
- 🗂️ **Record arrays**:  
  - Allow mixing different data types in one array, with **named fields**.  
  - Example: David Bowie discography with fields for title (`U32` string), release date (`M8[D]` datetime), and chart rank (`int64`).  
  - Fields can be accessed by index (like tuples) or by name (like dicts).  
  - Field lengths and data type descriptors are important because data is stored contiguously in memory.  
  - Useful for structured datasets, though Pandas DataFrames are often more powerful for real-world work.  

- 📅 **Datetime objects (`datetime64`)**:  
  - Encode dates and times with granularity from days down to nanoseconds.  
  - Can be initialized with strings of varying detail (year, full date, date+time).  
  - Support comparisons (`<`, `>`) and arithmetic (subtraction → `timedelta`).  
  - Work seamlessly with NumPy functions like `np.diff` to compute intervals between events.  
  - Can be used in ranges (`np.arange`) for sequences of dates.  

- ⚡ **Practical applications**:  
  - Record arrays are handy for small structured datasets where you want field names and mixed types.  
  - Datetime64 objects make time series analysis efficient, and their functionality extends further in **Pandas**, which builds on NumPy.  

👉 In short: Record arrays give you **structured, labeled storage** inside NumPy, while datetime64 provides **powerful date/time handling**. Together, they bridge numerical computing with structured and temporal data, paving the way toward Pandas DataFrames.

Filling missing values

- 🔄 **Generalization**:

Challenge: Unisex baby names
---
**🛠️ Step‑by‑step workflow**

1. **Group totals by name and sex**  
   ```python
   totals = names.groupby(["name","sex"])["count"].sum().unstack()
   ```
   - This gives a DataFrame with names as rows, and two columns: `F` and `M` (female and male totals).  
   - Some names may appear only in one column → you’ll see NaNs.

2. **Drop NaNs**  
   ```python
   totals = totals.dropna()
   ```
   - Keeps only names that appear in both male and female lists.

3. **Compute ratio**  
   ```python
   ratio = totals["M"] / totals["F"]
   ```
   - A name is “unisex” if `0.5 <= ratio <= 2`.

4. **Filter unisex names**  
   ```python
   unisex = totals[(ratio >= 0.5) & (ratio <= 2)]
   ```

5. **Find top 10 unisex names**  
   ```python
   top10_unisex = unisex.sum(axis=1).sort_values(ascending=False).head(10)
   print(top10_unisex)
   ```
   - This gives the most common unisex names across all years.

6. **Plot popularity trends**  
   - For each of the top 10 names, plot male vs female counts across years:  
     ```python
     import matplotlib.pyplot as plt

     for name in top10_unisex.index:
         subset = names[names["name"] == name]
         subset.groupby(["year","sex"])["count"].sum().unstack().plot(title=name)
         plt.show()
     ```
   - Each plot shows how the name’s usage evolved for boys vs girls.

**🎯 What you’ll see**
- Names like **Taylor, Jordan, Casey, Alex, Morgan, Jamie** often emerge as top unisex names.  
- Some start male‑dominant and shift female (or vice versa).  
- The ratio filter ensures they’re balanced within a factor of two overall.

👉 This challenge combines **groupby, unstack, ratio filtering, and plotting** — a perfect exercise in Pandas fluency.
