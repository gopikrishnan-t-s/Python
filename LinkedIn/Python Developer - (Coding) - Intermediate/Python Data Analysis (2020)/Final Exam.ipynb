   - Also correct. `.assign()` creates a new DataFrame with the added column.

The option that is **NOT** a proper way is:

**`df.y = f(df.x)`**

👉 In short: always prefer `df['y'] = ...` or `df.assign(...)` for clarity and reliability. Attribute assignment (`df.y = ...`) can work but is less safe, and chaining it after creating a dummy column is not recommended. 

The question is:

**In data type specifications, to what does the `<` sign refer?**

**🔎 Explanation**

In **NumPy data type specifications**, symbols like `<`, `>`, `=`, and `|` indicate **byte order** (endianness):

- `<` → **little‑endian** (least significant byte stored first).  
- `>` → **big‑endian** (most significant byte stored first).  
- `=` → native byte order of the machine.  
- `|` → not applicable (no byte order, e.g., for 1‑byte types).

So when you see something like `dtype('<i4')`, it means:
- Integer type (`i`)  
- 4 bytes (`4`)  
- Stored in **little‑endian** order (`<`).

**✅ Correct Answer**

**The `<` sign refers to the byte order (little‑endian).**
