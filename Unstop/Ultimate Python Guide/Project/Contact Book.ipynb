## Introduction to CSV:
CSV (Comma Separated Values) is a simple file format used to store tabular data. It works like a simplified spreadsheet.
* Each line represents a row
* Each value is separated by a comma
* Easy to store and exchange data between applications
### Key Features of CSV:
* **Simplicity**: Plain text format, easy to read and create
* **Universality**: Compatible with spreadsheets, databases, and applications
* **Efficiency**: Lightweight and fast for storing and transferring data
### Example Table:
| Name | Phone | Email |
| --- | --- | --- |
| John Doe | 123-456-7890 | john.doe@example.com |
| Jane Smith | 987-654-3210 | jane.smith@example.com |
### Same Data in CSV Format:

## Code Example: Adding a Contact
This function adds a new contact to the CSV file.

````python
import csv

def add_contact(filename, contact):
    with open(filename, 'a', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(contact)
````

#### Explanation:
* `import csv` Imports the CSV module.
* `add_contact(filename, contact)` Creates a function that takes:
    * filename
    * contact list
* `open(filename, 'a', newline='')` Opens file in append mode to add new data without deleting old data.
* `csv.writer(csvfile)` Creates a writer object.
* `writer.writerow(contact)` Writes contact data as a new row.

## Code Example: Viewing Contacts
This function reads and displays all contacts.

````python
def view_contacts(filename):
    with open(filename, 'r') as csvfile:
        reader = csv.reader(csvfile)
        for row in reader:
            print(row)
````

#### Explanation:
* Opens CSV file in read mode
* Uses `csv.reader()` to read rows
* Loops through each row
* Prints each contact

## Code Example: Searching for a Contact
This function searches for a contact by name.

````python
def search_contact(filename, search_term):
    with open(filename, 'r') as csvfile:
        reader = csv.reader(csvfile)
        for row in reader:
            if search_term in row[0]:  # name in first column
                print(row)
````

#### Explanation:
* Opens file in read mode
* Reads each row
* Checks if search term matches name column
* Prints matching contacts
