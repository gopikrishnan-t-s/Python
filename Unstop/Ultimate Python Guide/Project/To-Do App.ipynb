# **Project: To-Do App – Building Your First Task Manager**

## What is a To-Do App?
A **To-Do App** is a digital tool designed to help users manage and organize their tasks. It is essentially a digital version of a notepad with enhanced features that improve productivity and efficiency.

You can think of it as a personal assistant that helps track everything from simple grocery lists to complex project deadlines.

### Key Benefits:
* **Organization**: Keeps tasks neatly organized
* **Prioritization**: Allows setting importance levels
* **Time Management**: Helps manage time effectively
* **Reminders**: Sends notifications to keep you on track

## Key Features of a To-Do App
A typical To-Do App includes the following core features:
* **Task Creation**: Add new tasks with title and description
* **Status Updates**: Mark tasks as "To-Do", "In Progress", or "Completed"
* **Deadline Management**: Set due dates and reminders

These features help streamline daily activities and improve productivity.

## Core Functionality:
### 1. Task Creation and Status Updates:
This includes:
* Adding tasks with titles and descriptions
* Updating task status from:
    * To-Do
    * In Progress
    * Completed
### 2. Deadline Management:
This includes:
* Setting due dates for tasks
* Adding reminders to ensure timely completion

## Code Example: Task Class – Part 1
We will use Python to build our To-Do App due to its simplicity and rich library support.

```python
class Task:
    def __init__(self, title, description, due_date):
        self.title = title
        self.description = description
        self.due_date = due_date
        self.status = "to-do"
```

#### Explanation:
* We define a class called `Task` that acts as a blueprint for individual tasks.
* The `__init__` method initializes each task.
* It accepts:
    * `title`
    * `description`
    * `due_date`
* Each value is stored using instance variables.
* The default status is set to "`to-do`" for every new task.

This creates the foundation for managing tasks in the application.

## Code Example: Task Class – Part 2
Now we add more functionality to the class.

```python
def update_status(self, new_status):
    self.status = new_status

def __str__(self):
    return f"{self.title} - {self.status} - {self.due_date}"
```

#### Explanation:
* `update_status(new_status)` Updates the task status (for example: to-do → in progress → completed).
* `__str__()` Defines how the task will appear when printed. It returns a formatted string containing:
    * Task title
    * Current status
    * Due date

This makes task display clear and user-friendly.
