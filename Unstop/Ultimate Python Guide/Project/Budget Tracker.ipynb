# **Project: Budget Tracker**

## What is a Budget Tracker?
A **budget tracker** is a tool that helps you manage your money.

It records:
* **Income** (money coming in)
* **Expenses** (money going out)
* **Savings** (money left)

It gives you a clear view of your finances and helps you save better.

## Key Components of a Budget Tracker:
Every budget tracker has three main parts:
### Income:
Money you receive:
* Salary
* Freelance income
* Investments
### Expenses:
Money you spend:
* Rent
* Food
* Travel
* Entertainment
### Savings:
Money left after spending.
#### Formula:
$\text{Savings} = \text{Income} - \text{Expenses}$

## Coding a Budget Tracker in Python:
We will build a simple budget tracker step by step.
### Part 1: Transaction Class
This class stores details of each transaction.

```python
class Transaction:
    def __init__(self, date, description, amount, type):
        self.date = date
        self.description = description
        self.amount = amount
        self.type = type
```

#### What it stores:
* Date
* Description
* Amount
* Type (income/expense)

### Part 2: Budget Class:
This class manages all transactions.

```python
class Budget:
    def __init__(self):
        self.transactions = []

    def add_transaction(self, transaction):
        self.transactions.append(transaction)

    def calculate_income(self):
        return sum(t.amount for t in self.transactions if t.type == 'income')

    def calculate_expenses(self):
        return sum(t.amount for t in self.transactions if t.type == 'expense')

    def calculate_savings(self):
        return self.calculate_income() - self.calculate_expenses()
```

#### Functions:
* Add transaction
* Calculate income
* Calculate expenses
* Calculate savings

### Part 3: Testing the Budget Tracker
Let’s test it.

```python
from datetime import date

budget = Budget()

budget.add_transaction(Transaction(date(2024, 1, 15), 'Salary', 5000, 'income'))
budget.add_transaction(Transaction(date(2024, 1, 20), 'Rent', 1500, 'expense'))

income = budget.calculate_income()
expenses = budget.calculate_expenses()
savings = budget.calculate_savings()

print(f'Income: ${income}')
print(f'Expenses: ${expenses}')
print(f'Savings: ${savings}')
```
